```python

import pandas as pd
import plotly.express as px

# Load dataset
df = pd.read_csv('earthquake_1995-2023.csv')

# Before cleaning
print("Dataset Shape Before Cleaning:", df.shape)
print("\nMissing Values Before Cleaning:")
print(df.isnull().sum())
print("\nDuplicate Rows Before Cleaning:", df.duplicated().sum())

# Data Cleaning

# Convert date_time to datetime
df['date_time'] = pd.to_datetime(
    df['date_time'],
    format='%d/%m/%Y %H:%M'
)

# Extract year
df['year'] = df['date_time'].dt.year

# Fill missing alert values
df['alert'] = df['alert'].fillna('none')

# Remove duplicate rows
df = df.drop_duplicates().reset_index(drop=True)

# After cleaning
print("\nDataset Shape After Cleaning:", df.shape)
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
print("\nDuplicate Rows After Cleaning:", df.duplicated().sum())

# Alert levels
print("\nAlert Level Counts:")
print(df['alert'].value_counts())

# Statistical summary
print("\nStatistical Summary:")
display(df[['magnitude', 'depth', 'sig']].describe())

# Display cleaned dataset
print("\nCleaned Dataset:")
display(df.head())

# Save cleaned dataset
df.to_csv('earthquake_cleaned.csv', index=False)

print("\nData Cleaning Completed Successfully!")
```

```python
# Task 1: Core Interactive Charts

import pandas as pd
import plotly.express as px

# Load cleaned dataset
df = pd.read_csv('earthquake_cleaned.csv')

# Ensure correct date format
df['date_time'] = pd.to_datetime(df['date_time'])

# Fill missing alert values
df['alert'] = df['alert'].fillna('none')

# Chart 1: Interactive Histogram
fig1 = px.histogram(
    df,
    x='magnitude',
    nbins=30,
    title='Earthquake Magnitude Distribution'
)

fig1.update_layout(xaxis_rangeslider_visible=True)
fig1.show()


# Chart 2: Scatter Plot (Magnitude vs Depth)
fig2 = px.scatter(
    df,
    x='magnitude',
    y='depth',
    color='alert',
    hover_data=['location', 'country', 'date_time'],
    log_y=True,
    title='Magnitude vs Depth by Alert Level'
)

fig2.show()


# Chart 3: Box Plot by Alert Level
fig3 = px.box(
    df,
    x='alert',
    y='magnitude',
    color='alert',
    points='all',
    category_orders={
        'alert': ['green', 'yellow', 'orange', 'red', 'none']
    },
    title='Magnitude Distribution by Alert Level'
)

fig3.show()


# Chart 4: 3D Scatter Plot
fig4 = px.scatter_3d(
    df,
    x='magnitude',
    y='depth',
    z='sig',
    color='alert',
    hover_data=['location'],
    title='3D Earthquake Analysis'
)

fig4.show()
```

```python
# Task 2: Animation and Dropdown Controls

import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Load cleaned dataset
df = pd.read_csv('earthquake_cleaned.csv')

# Convert date and extract year
df['date_time'] = pd.to_datetime(df['date_time'])
df['year'] = df['date_time'].dt.year

# Fill missing alert values
df['alert'] = df['alert'].fillna('none')

# Sort by year
df = df.sort_values('year')

# TASK 2 - CHART 5: ANIMATED SCATTER PLOT

fig5 = px.scatter(
    df,
    x='magnitude',
    y='depth',
    size='sig',
    color='alert',
    animation_frame='year',
    hover_data=['location', 'date_time'],
    range_x=[6, 10],
    range_y=[0, 700],
    size_max=35,
    category_orders={
        'alert': ['green', 'yellow', 'orange', 'red', 'none']
    },
    title='Animated Earthquake Analysis (1995-2023)',
    labels={
        'magnitude': 'Magnitude',
        'depth': 'Depth (km)',
        'sig': 'Significance Score',
        'alert': 'Alert Level'
    }
)

# Set animation speed
if fig5.layout.updatemenus:
    fig5.layout.updatemenus[0].buttons[0].args[1]['frame']['duration'] = 700

fig5.show()


# TASK 2 - CHART 6: DROPDOWN SCATTER PLOT

fig6 = go.Figure()

metrics = ['depth', 'sig', 'mmi']

# Create traces for each metric
for i, metric in enumerate(metrics):
    fig6.add_trace(
        go.Scatter(
            x=df['magnitude'],
            y=df[metric],
            mode='markers',
            name=metric,
            visible=(i == 0),
            marker=dict(
                size=7,
                opacity=0.6
            )
        )
    )

# Create dropdown buttons
buttons = []

for i, metric in enumerate(metrics):
    visible = [False] * len(metrics)
    visible[i] = True

    buttons.append(
        dict(
            label=metric.upper(),
            method='update',
            args=[
                {'visible': visible},
                {
                    'yaxis': {
                        'title': metric.upper(),
                        'autorange': True
                    }
                }
            ]
        )
    )

# Update chart layout
fig6.update_layout(
    title='Magnitude vs Earthquake Metrics',
    xaxis_title='Magnitude',
    yaxis_title='Depth (km)',
    updatemenus=[
        dict(
            buttons=buttons,
            direction='down',
            x=1.05,
            y=1.15,
            showactive=True
        )
    ]
)

fig6.show()


# COMPARISON:
# The animated scatter plot shows how earthquake
# magnitude, depth and significance change yearly.
# A single static frame shows only one year's data
# and cannot display temporal changes.
```
